# CarCare AI exploratory analysis
This notebook examines the reproducible synthetic dataset. It is not real-world vehicle-service data.

```python
from config import DATA_PATH
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df.shape, df.isna().sum().sum(), df.duplicated().sum()
```

```python
df.describe(include='all').T
```

```python
sns.histplot(df, x='current_odometer_km', bins=35); plt.title('Synthetic car odometer distribution'); plt.show()
```

```python
sns.boxplot(data=df, x='driving_condition', y='remaining_km_to_service'); plt.title('Synthetic remaining service distance by driving condition'); plt.show()
```

```python
sns.heatmap(df.select_dtypes('number').corr(), cmap='vlag', center=0); plt.title('Numeric-feature correlation'); plt.show()
```

The plots describe generator assumptions. Do not generalize these distributions or model metrics to real vehicles.
